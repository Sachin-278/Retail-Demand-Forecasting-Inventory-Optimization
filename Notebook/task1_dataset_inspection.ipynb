# DATA CLEANING

```python
calendar["date"] = pd.to_datetime(calendar["date"])
```

```python
calendar["date"].dtype
```

```python
calendar[["event_name_1", "event_type_1", "event_name_2", "event_type_2"]].isnull().sum()
```

```python
event_cols = [
    "event_name_1",
    "event_type_1",
    "event_name_2",
    "event_type_2"
]

calendar[event_cols] = calendar[event_cols].fillna("No Event")
```

## Data Quality Validation

The M5 datasets were checked for missing values, invalid sales and price values, and duplicate records.

```python
# Sales validation
daily_sales = sales.filter(regex=r"^d_\d+$")

print("Sales missing values:", sales.isnull().sum().sum())
print("Negative sales values:", (daily_sales < 0).sum().sum())
print("Duplicate sales IDs:", sales["id"].duplicated().sum())
```

```python
# Calendar validation

print("Calendar missing values:", calendar.isnull().sum().sum())
print("Duplicate calendar dates:", calendar["date"].duplicated().sum())
```

```python
# Sell Prices validation

print("Price missing values:", prices["sell_price"].isnull().sum())
print("Zero/negative prices:", (prices["sell_price"] <= 0).sum())
print("Duplicate price records:", prices.duplicated(["store_id", "item_id", "wm_yr_wk"]).sum())
```
